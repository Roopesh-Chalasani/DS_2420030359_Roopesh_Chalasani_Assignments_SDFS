```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.datasets import load_iris
iris=load_iris()
df=pd.DataFrame(data=iris.data, columns=iris.feature_names)
df['species']=iris.target
plt.figure(figsize=(6,4))
sns.histplot(df['sepal length (cm)'], bins=20, kde=True,color='orange')
plt.title("Histogram of Sepal Length")
plt.show()

#boxplot
plt.figure(figsize=(6,4))
sns.boxplot(x=df['sepal length (cm)'])
plt.title("Boxplot of Sepal Length")
plt.show()

#pie-chart
species_counts = df['species'].value_counts()
plt.figure(figsize=(6,6))
plt.pie(species_counts,labels=iris.target_names,autopct='%1.1f%%')
plt.title("Pie Chart of Species Distribution")
plt.show()


```

```python
#2.Bivarriate visualizations
#scatter plot
plt.figure(figsize=(6,4))
sns.scatterplot(x=df['sepal length (cm)'], y=df['sepal width (cm)'], hue=df['species'], palette='Set1')
plt.title("Scatter Plot of Sepal Length vs Sepal Width")    
plt.show()

#line-chart
plt.figure(figsize=(6,4))
sns.lineplot(x=df['sepal length (cm)'], y=df['sepal width (cm)'],hue=df['species'])
plt.title("Line Chart of Sepal Length vs Sepal Width")
plt.show()

#Bar-chart
plt.figure(figsize=(6,4))
sns.barplot(x=df['species'], y=df['sepal length (cm)'], color='brown')
plt.title("Bar Chart of mean Sepal Length by Species")
plt.show()
```

```python
#multivariate visualizations
#heatmap(correlation matrix)
plt.figure(figsize=(6,4))
sns.heatmap(df.iloc[:,:4].corr(), annot=True, cmap='coolwarm')
plt.title("Heatmap of Feature correlations")
plt.show()

#Bubble chart (scatterplot with size=petal length)
plt.figure(figsize=(6,4))
plt.scatter(df['sepal length (cm)'], df['sepal width (cm)'], s=df['petal length (cm)']*20, c=df['species'], alpha=0.5)
plt.title("Bubble Chart: Sepal vs petal")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Sepal Width (cm)")
plt.show()
# pair plot (relationships among all features)
sns.pairplot(df.iloc[:,:4])
plt.suptitle("Pair plot of iris Features",y=1.02)
plt.show()
```

```python
#4. Advanced Techniques
#violin plot
plt.figure(figsize=(6,2))
sns.violinplot(x=df['species'],y=df['sepal length (cm)'])
plt.title("Violin plot: sepal length by species")
plt.xticks([0, 1, 2], iris.target_names)
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Load tips dataset
tips = sns.load_dataset('tips')

# Display first few rows
print(tips.head())

# Histogram
plt.figure(figsize=(6,4))
sns.histplot(tips['total_bill'], bins=20, kde=True, color='orange')
plt.title("Histogram of Total Bill")
plt.show()

# Boxplot
plt.figure(figsize=(6,4))
sns.boxplot(x=tips['total_bill'])
plt.title("Boxplot of Total Bill")
plt.show()

# Pie Chart
day_counts = tips['day'].value_counts()

plt.figure(figsize=(6,6))
plt.pie(day_counts, labels=day_counts.index, autopct='%1.1f%%')
plt.title("Pie Chart of Customers by Day")
plt.show()
```

```python
#2.Bivarriate visualizations
#scatter plot
plt.figure(figsize=(6,4))
sns.scatterplot(x=tips['total_bill'], y=tips['tip'], hue=tips['day'], palette='Set1')
plt.title("Scatter Plot of Total Bill vs Tip")     
plt.show()
#line-chart
plt.figure(figsize=(6,4))
sns.lineplot(x=tips['total_bill'], y=tips['tip'],hue=tips['day'])
plt.title("Line Chart of Total Bill vs Tip")
plt.show()
#Bar-chart
plt.figure(figsize=(6,4))
sns.barplot(x=tips['day'], y=tips['total_bill'], color='brown')
plt.title("Bar Chart of mean Total Bill by Day")
plt.show()
```

```python
#multivariate visualizations
#heatmap(correlation matrix)
plt.figure(figsize=(6,4))
sns.heatmap(tips.select_dtypes(include='number').corr(), annot=True, cmap='coolwarm')
plt.title("Heatmap of Feature correlations")
plt.show()

#Bubble chart (scatterplot with size=petal length)
plt.figure(figsize=(6,4))
plt.scatter(tips['total_bill'], tips['tip'], s=tips['size']*20, c=tips['smoker'].astype('category').cat.codes, alpha=0.5)
plt.title("Bubble Chart: Total Bill vs Tip")
plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.show()

# pair plot (relationships among all features)
sns.pairplot(tips.iloc[:,:4])
plt.suptitle("Pair plot of Tips Features",y=1.02)
plt.show()
```

```python
#4. Advanced Techniques
#violin plot
plt.figure(figsize=(6,2))
sns.violinplot(x=tips['day'],y=tips['total_bill'])
plt.title("Violin plot: total bill by day")
plt.xticks([0,1,2], ['Thur','Fri','Sat'])
plt.show()
```
